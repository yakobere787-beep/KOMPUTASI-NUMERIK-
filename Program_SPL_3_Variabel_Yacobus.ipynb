```python
# Memanggil library numpy
import numpy as np

# 1. Memasukkan matriks koefisien A
A = np.array([
    [2, 1, -1],
    [-1, 3, 2],
    [3, -2, 4]
], dtype=float)

# 2. Memasukkan vektor konstanta B
B = np.array([8, 9, 3], dtype=float)

# 3. Membentuk matriks augmented
M = np.column_stack((A, B))

# 4. Menentukan jumlah variabel
n = len(B)

print("Matriks Augmented:")
print(M)

# 5. Proses Eliminasi Gauss
for k in range(n):

    # Memilih elemen pivot terbesar
    pivot = k + np.argmax(np.abs(M[k:, k]))

    # Menukar baris jika diperlukan
    M[[k, pivot]] = M[[pivot, k]]

    # Eliminasi elemen di bawah pivot
    for i in range(k + 1, n):
        faktor = M[i, k] / M[k, k]
        M[i, k:] = M[i, k:] - faktor * M[k, k:]

print("\nMatriks Setelah Eliminasi:")
print(M)

# 6. Proses Substitusi Mundur
X = np.zeros(n)

for i in range(n - 1, -1, -1):
    X[i] = (
        M[i, n] - np.dot(M[i, i+1:n], X[i+1:n])
    ) / M[i, i]

# 7. Menampilkan Hasil Penyelesaian
print("\nHasil Penyelesaian SPL:")
print("x =", X[0])
print("y =", X[1])
print("z =", X[2])

# 8. Verifikasi Hasil
hasil = A @ X
error = np.linalg.norm(hasil - B)

print("\nVerifikasi Hasil:")
print("A * X =", hasil)
print("B     =", B)
print("Error =", error)

print("\nProgram Selesai")
```
